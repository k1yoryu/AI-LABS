```python
import pandas as pd

df=pd.read_csv("water_potability.csv")
df.head(15)
```

```python
propuski_po_stolbcam=df.isnull()
print(propuski_po_stolbcam.sum())
```

```python
ph_mediana=df["ph"].median()
sulfate_mean=df["Sulfate"].mean()
trihal_mode=df["Trihalomethanes"].mode()[0]

df["ph"]=df["ph"].fillna(ph_mediana)
df["Sulfate"]=df["Sulfate"].fillna(sulfate_mean)
df["Trihalomethanes"]=df["Trihalomethanes"].fillna(trihal_mode)
df.to_csv("water_potability_new.csv", index=False)
print(df.isnull().sum())
```
